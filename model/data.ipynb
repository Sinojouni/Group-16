```python
import pandas as pd
import numpy as np
import math, ast

def parse_json_list(x, key="name", topn=None):
    if isinstance(x, list):
        values = [d.get(key, d) if isinstance(d, dict) else d for d in x]
        return values[:topn] if topn else values
    try:
        data = ast.literal_eval(x)
        if isinstance(data, list):
            values = [d.get(key, d) if isinstance(d, dict) else d for d in data]
            return values[:topn] if topn else values
    except Exception:
        return []
    return []

```

```python
df["genres"]   = df["genres"].apply(lambda x: parse_json_list(x))
df["cast"]     = df["cast"].apply(lambda x: parse_json_list(x, topn=3))
df["keywords"] = df["keywords"].apply(lambda x: parse_json_list(x))

df["director"] = df["crew"].apply(
    lambda x: [d.get("name") for d in ast.literal_eval(x) if isinstance(d, dict) and d.get("job")=="Director"]
    if isinstance(x, str) else
    [d.get("name") for d in x if isinstance(d, dict) and d.get("job")=="Director"]
)

df = df.drop(columns=["crew"])

df["movie_id"] = df["tmdbId"].astype(str)
```

```python
#code from chat gpt to get en efficent simple from the dataset so we take the sample based on the nb of users
#here (12k user)
import numpy as np, pandas as pd, math, os

MAX_USERS          = 17_000
MIN_USER_RATINGS   = 10
MIN_ITEM_RATINGS   = 5
VAL_FRACTION       = 0.15
SEED               = 42
np.random.seed(SEED)

interactions = df[["userId", "movie_id", "rating", "tmdbId", "genres", "cast", "keywords", "director", "overview", "original_language", "runtime", "popularity", "vote_average", "vote_count", "release_year", "title"]].copy()

interactions["rating"] = interactions["rating"].astype(float)

user_activity = interactions.groupby("userId").size().rename("n_ratings")
active_users = user_activity[user_activity >= MIN_USER_RATINGS]

quantiles = active_users.quantile([0.5, 0.8, 0.95]).to_dict()

q50 = active_users[active_users <= quantiles[0.5]]
q80 = active_users[(active_users > quantiles[0.5]) & (active_users <= quantiles[0.8])]
q95 = active_users[(active_users > quantiles[0.8]) & (active_users <= quantiles[0.95])]
q99 = active_users[active_users > quantiles[0.95]]

alloc = {
    "q50": int(MAX_USERS * 0.55),
    "q80": int(MAX_USERS * 0.25),
    "q95": int(MAX_USERS * 0.12),
    "q99": int(MAX_USERS * 0.08),
}

def sample_index(series, k):
    if len(series) <= k:
        return series.index.tolist()
    return np.random.choice(series.index, size=k, replace=False).tolist()

selected_users = []
selected_users += sample_index(q50, alloc["q50"])
selected_users += sample_index(q80, alloc["q80"])
selected_users += sample_index(q95, alloc["q95"])
selected_users += sample_index(q99, alloc["q99"])
selected_users = list(dict.fromkeys(selected_users))

if len(selected_users) > MAX_USERS:
    selected_users = selected_users[:MAX_USERS]

sampled = interactions[interactions.userId.isin(selected_users)].copy()

item_activity = sampled.groupby("movie_id").size().rename("item_interactions")
valid_items = item_activity[item_activity >= MIN_ITEM_RATINGS].index
sampled = sampled[sampled.movie_id.isin(valid_items)]

sampled = sampled.sort_values(["userId", "movie_id"]).reset_index(drop=True)

unique_users = sampled.userId.unique()
unique_items = sampled.movie_id.unique()

user2idx = {u: i for i, u in enumerate(unique_users)}
item2idx = {m: i for i, m in enumerate(unique_items)}

sampled["user_idx"] = sampled.userId.map(user2idx).astype("int32")
sampled["item_idx"] = sampled.movie_id.map(item2idx).astype("int32")


train_rows = []
val_rows = []
for uid, grp in sampled.groupby("user_idx", sort=False):
    n = len(grp)
    if n <= 1:
        train_rows.append(grp)
        continue
    val_size = max(1, int(math.floor(n * VAL_FRACTION)))
    val_idx = np.random.choice(grp.index, size=val_size, replace=False)
    val_rows.append(grp.loc[val_idx])
    train_rows.append(grp.drop(val_idx))

train_df = pd.concat(train_rows).sort_index().reset_index(drop=True)
val_df   = pd.concat(val_rows).sort_index().reset_index(drop=True)

print("Sampled users:", len(unique_users))
print("Sampled items:", len(unique_items))
print("Train interactions:", len(train_df))
print("Val interactions:", len(val_df))
print("Overall sparsity: {:.4f}".format(1 - (len(sampled) / (len(unique_users) * len(unique_items)))))


print("Avg train ratings per user:", train_df.groupby("user_idx").size().mean())
print("Avg train ratings per item:", train_df.groupby("item_idx").size().mean())

sampled_df = sampled

sampled.head()
```
